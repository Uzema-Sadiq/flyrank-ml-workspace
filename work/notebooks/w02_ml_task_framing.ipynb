## 1. My lane as an ML task (type)

**Lane:** Content Refresh Prioritization

**Task Type:** Classification

This project is a classification problem because the goal is to predict whether a webpage should be prioritized for content refresh using observable SEO signals. Each webpage is classified as either declining or not declining based on its trend. The model supports decision-making by helping identify pages that may require review before they lose more traffic.

```python
import pandas as pd

df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

print('Distribution of trend_direction:')
print(df['trend_direction'].value_counts())
```

## 2. Target or proxy

The prediction target is whether a webpage is declining.

The starter dataset provides the `trend_direction` column, which can be used as a proxy target. Pages with the value `down` are treated as declining pages that may benefit from content refresh.

```python
df['is_declining'] = (df['trend_direction'].str.lower() == 'down').astype(int)

print(df[['trend_direction', 'is_declining']].head())
```

## 3. Success metric

The primary success metric is **Precision@50**.

Precision@50 measures how many of the top 50 webpages recommended by the model are actually declining. A higher Precision@50 means the model provides more useful recommendations for prioritizing content refresh and helps reduce unnecessary manual review.

```python
print('Positive class rate:')
print(df['is_declining'].mean())
```

## 4. The unit of analysis, as a real dataframe

The unit of analysis is **one webpage**.

Each row in the dataset represents a single webpage and contains observable SEO features such as impressions, CTR, average position, search volume, content age, and update history. These features can be used to predict whether a page should be prioritized for refresh.

```python
print('Dataset shape:', df.shape)

print('\nColumns:')
print(df.columns.tolist())

print('\nFirst five rows:')
df.head()
```

## 5. Why ML beats a fixed rule here

A fixed rule usually depends on one or two manually selected thresholds, such as content age or impressions.

Machine learning can combine multiple signals at the same time and identify relationships that are difficult to capture with a simple rule. This supports more consistent and scalable prioritization when working with thousands of webpages. The model provides decision support rather than guaranteeing future outcomes.

```python
features = [
    'impressions_90d',
    'ctr',
    'content_age_days',
    'avg_position',
    'days_since_last_update'
]

print('Example features used for prediction:')
print(features)
```
